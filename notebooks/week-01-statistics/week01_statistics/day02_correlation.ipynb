calculate covarience

```python
import pandas as pd

data = {
    'job_role': [
        'Data Analyst',
        'Data Scientist',
        'ML Engineer',
        'Data Analyst',
        'Data Scientist',
        'ML Engineer'
    ],
    'salary_lpa': [4, 8, 12, 5, 10, 15],
    'experience_years': [0, 2, 3, 1, 4, 5]
}

df = pd.DataFrame(data)

print(df)
```

```python
covariance = df['experience_years'].cov(
    df['salary_lpa']
)
    


print("Covariance:", covariance)
```

```python
correlation = df['experience_years'].corr(
    df['salary_lpa']
)

print("Correlation:", correlation)
```

```python
correlation_matrix = df[
    ['experience_years', 'salary_lpa']
].corr()

print(correlation_matrix)
```
