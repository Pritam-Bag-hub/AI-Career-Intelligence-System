```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("synthetic_job_market.csv")
```

```python
df.head()
```

```python
role_analysis = df.groupby("job_role")["salary_lpa"].agg(
    ["count", "mean", "median", "skew"]
)

print(role_analysis)
```

```python
plt.figure(figsize=(11, 6))

sns.boxplot(
    data=df,
    x="job_role",
    y="salary_lpa"
)

plt.title("Salary Distribution Across Job Roles")
plt.xlabel("Job Role")
plt.ylabel("Salary (LPA)")
plt.xticks(rotation=15)

plt.show()
```
