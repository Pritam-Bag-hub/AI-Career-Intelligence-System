```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    'job_role': [
        'Data Analyst', 'Data Scientist', 'ML Engineer',
        'Data Analyst', 'Data Scientist', 'ML Engineer',
        'Data Analyst', 'Data Scientist', 'ML Engineer',
        'Data Analyst', 'Data Scientist', 'ML Engineer'
    ],
    'salary_lpa': [4, 8, 12, 5, 10, 15, 6, 11, 16, 7, 13, 19],
    'experience_years': [0, 2, 3, 1, 4, 5, 2, 5, 6, 3, 6, 8],
    'skills_count': [4, 7, 9, 5, 8, 10, 6, 9, 11, 7, 10, 13],
    'company_size': [50, 200, 500, 80, 300, 700,
                     100, 400, 900, 150, 600, 1200]
}

df = pd.DataFrame(data)

print(df.head())
```

```python
pearson_corr = df['experience_years'].corr(
    df['salary_lpa'],
    method = 'pearson'
)

spearman_corr = df['experience_years'].corr(
    df['salary_lpa'],
    method = 'spearman'
)

print("Pearson:", pearson_corr)
print("Spearman:", spearman_corr)
```

```python
sns.scatterplot(
    data=df,
    x='experience_years',
    y='salary_lpa'
)

plt.title('Experience vs Salary')
plt.xlabel('Experience (Years)')
plt.ylabel('Salary (LPA)')
plt.show()
```

```python
correlation_matrix = df.select_dtypes(
    include='number'
).corr()

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Job Market Correlation Heatmap')
plt.show()
```

```python
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    vmin=-1,
    vmax=1,
    center=0
)

plt.title('Job Market Correlation Heatmap')
plt.show()
```
