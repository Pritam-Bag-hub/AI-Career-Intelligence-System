```python
import pandas as pd

data = {
    'job_role': [
        'Data Analyst',
        'Data Scientist',
        'ML Engineer',
        'Data Analyst',
        'Data Scientist',
        'ML Engineer'
    ],
    'salary_lpa': [4, 8, 12, 5, 10, 15],
    'experience_years': [0, 2, 3, 1, 4, 5]
}

df = pd.DataFrame(data)
```

```python
mean_salary = df['salary_lpa'].mean()
print("Mean Salary:", mean_salary)
```

```python
median_salary = df['salary_lpa'].median()
print("median salary :", median_salary)
```

```python
standard_deviation_salary = df['salary_lpa'].std()
print("standard deviation", standard_deviation_salary)
```

```python
minimum_salary = df['salary_lpa'].min()
maximum_salary = df['salary_lpa'].max()
print("minimum salary :", minimum_salary)
print("maximum salary :", maximum_salary)
```

```python
avg_salary_of_each_jobrole = df.groupby('job_role')['salary_lpa'].mean()
print("avg salary of each job role :", avg_salary_of_each_jobrole)
```

```python
q25_salary = df['salary_lpa'].quantile(0.25)
print("25th Percentile Salary:", q25_salary)
```

```python
q50_salary = df['salary_lpa'].quantile(0.50)
print("50th Percentile Salary:", q50_salary)
```

```python
q75_salary = df['salary_lpa'].quantile(0.75)
print("75th Percentile Salary:", q75_salary)
```

```python
Q1 = df['salary_lpa'].quantile(0.25)
Q2 = df['salary_lpa'].quantile(0.50)
Q3 = df['salary_lpa'].quantile(0.75)
IQR = Q3-Q1
lower_bound = Q1 - (1.5*IQR)
upper_bound = Q3 + (1.5*IQR)
outlier = df[ (df['salary_lpa'] <lower_bound) |  (df['salary_lpa']>upper_bound)]
print("outlier:",outlier)
```
