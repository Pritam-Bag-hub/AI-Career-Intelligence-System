```python
import pandas as pd

data = {
    "job_role": [
        "Data Scientist",
        "Data Scientist",
        "Data Analyst",
        "Data Analyst"
    ],
    "python_required": [
        True,
        False,
        True,
        False
    ],
    "count": [40, 10, 20, 30]
}

df = pd.DataFrame(data)
print(df)
```

```python
df['count'].sum()
```

```python
df.groupby('job_role')['count'].sum()
```

```python
df[df['python_required'] == True]['count'].sum()
```

```python
python_jobs = df[df['python_required']]['count'].sum()

total_jobs = df['count'].sum()

probability = python_jobs / total_jobs

print(probability)
```

```python
data_science = df[df['job_role'] =='Data Scientist']['count'].sum()

total_jobs = df['count'].sum()

probability = data_science / total_jobs

print(probability)
```

```python
print(df[df['job_role'] == 'Data Scientist'])
```

```python
data_science = df[df['job_role'] == 'Data Scientist']['count'].sum()
print(data_science)
```

```python
total_jobs = df['count'].sum()
print(total_jobs)
```

```python
print(data_science / total_jobs)
```

```python
data_science_required_python = df[(df['job_role'] == 'Data Scientist')&(df['python_required'])]['count'].sum()
python_jobs = df[df['python_required']]['count'].sum()

probability_datascience_requiring_python = data_science_required_python / python_jobs
print(probability_datascience_requiring_python)
```
