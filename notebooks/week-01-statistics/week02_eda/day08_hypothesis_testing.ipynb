```python
import numpy as np
import pandas as pd

# 1. Set a random seed so your random numbers are reproducible every time you run the script
np.random.seed(42)

# Define how many rows (samples) you want
n_samples = 200

# 2. Simulate Experience Years (using an exponential distribution so most people have fewer years, and fewer people have high experience)
experience_years = np.random.exponential(scale=3, size=n_samples).round(1) + 1

# 3. Simulate whether Python is required (65% True, 35% False)
python_required = np.random.choice([True, False], size=n_samples, p=[0.65, 0.35])

# 4. Simulate Salary (making it dependent on experience and python requirement so there's a realistic pattern)
# Base salary = 5 LPA + (experience * 1.8) + (bonus if python is required) + random noise (bell curve)
salary_lpa = (
    5
    + (experience_years * 1.8)
    + (python_required * 3.5)
    + np.random.normal(loc=0, scale=2.5, size=n_samples)
)
salary_lpa = np.clip(
    salary_lpa, 3.5, 45.0
).round(2)  # Keep values realistic between 3.5 and 45 LPA

# 5. Simulate other categorical and numerical columns
job_roles = np.random.choice(
    ["Data Scientist", "Software Engineer", "Data Analyst", "ML Engineer"],
    size=n_samples,
    p=[0.3, 0.3, 0.2, 0.2],
)
company_size = np.random.choice(
    ["Startup", "Mid-size", "MNC"], size=n_samples, p=[0.4, 0.35, 0.25]
)
skill_count = np.random.randint(2, 9, size=n_samples)

# 6. Combine everything into a Pandas DataFrame
df_jobs = pd.DataFrame(
    {
        "job_role": job_roles,
        "experience_years": experience_years,
        "salary_lpa": salary_lpa,
        "skill_count": skill_count,
        "company_size": company_size,
        "python_required": python_required,
    }
)

# 7. Save to a CSV file so you can load it into your Jupyter Notebook anytime
df_jobs.to_csv("synthetic_job_market.csv", index=False)
print("Dataset created and saved successfully!")
print(df_jobs.head())
```

```python
python_salary = df_jobs[df_jobs['python_required']==True]["salary_lpa"]
```

```python
print(python_salary)
```

```python
non_python_salary = df_jobs[df_jobs["python_required"] == False]["salary_lpa"]
```

```python
print(non_python_salary)
```

```python
print("Python jobs:", python_salary.mean())
print("Non-Python jobs:", non_python_salary.mean())
```

```python
from scipy.stats import ttest_ind

result = ttest_ind(
    python_salary,
    non_python_salary,
    equal_var=False,
    alternative="greater"
)

print(result)
```

```python
p_value = result.pvalue

print("P-value:", p_value)
alpha = 0.05

if p_value < alpha:
    print("Reject the null hypothesis")
else:
    print("Fail to reject the null hypothesis")
```

```python
import numpy as np

np.random.seed(42)

# Simulated population of salaries
population = np.random.normal(
    loc=10,
    scale=4,
    size=100000
)

# Select a random sample
sample = np.random.choice(population, size=1000, replace=False)

# Calculate sample statistics
sample_mean = np.mean(sample)
sample_std = np.std(sample, ddof=1)

# Standard error
se = sample_std / np.sqrt(len(sample))

# 95% confidence interval
lower = sample_mean - 1.96 * se
upper = sample_mean + 1.96 * se

print("Population mean:", np.mean(population))
print("Sample mean:", sample_mean)
print("Standard error:", se)
print("95% CI:", lower, upper)
```
