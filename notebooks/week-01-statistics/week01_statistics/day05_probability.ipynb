```python
import numpy as np

salaries = np.array([5, 6, 7, 8, 9, 10, 11, 12, 13, 30])

mean = np.mean(salaries)
std = np.std(salaries)

z_scores = (salaries - mean) / std

print("Z-score for 30 from array:", z_scores[-1])
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

salaries = np.array([5, 6, 7, 8, 9, 10, 11, 12, 13, 100])

df = pd.DataFrame({'salary_lpa': salaries})

print("Mean:", df['salary_lpa'].mean())
print("Median:", df['salary_lpa'].median())
print("Standard deviation:", df['salary_lpa'].std())
```

```python
plt.hist(df['salary_lpa'], bins=5)
plt.xlabel("Salary (LPA)")
plt.ylabel("Frequency")
plt.title("Salary Distribution")
plt.show()
```
